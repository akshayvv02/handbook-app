```python
import os

def clean_folder_files(folder_path):
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)

        # Only process regular files, skip directories
        if os.path.isfile(file_path):
            with open(file_path, 'r', encoding='utf-8') as f:
                content = f.read()

            # Perform the replacements
            content = content.replace("{{ url_for('static', filename='", "")
            content = content.replace("') }}", "")

            # Write back the modified content
            with open(file_path, 'w', encoding='utf-8') as f:
                f.write(content)

    print("✔️ All file contents updated successfully.")
```

```python
# Example usage:
clean_folder_files('./html')  # Replace with your actual folder path
```
